# 3 向量
<a name="toc_40015_3.1"></a>
## 3.1 概述
<img align="right" src="../work/images/C1_W2_Lab04_Vectors.PNG" style="width:340px;" >在本课程中，向量是有序数值数组。在记号中，向量用小写粗体字母表示，例如 $\mathbf{x}$。向量中的元素具有相同类型。例如，一个向量不会同时包含字符和数值。数组中的元素数量通常称为*维度*，不过数学家可能更倾向于称其为*秩*。所示向量的维度为 $n$。向量元素可以通过索引引用。在数学语境中，索引编号通常从 1 到 n；在计算机科学和这些实验中，索引编号通常从 0 到 n-1。在记号中，单独引用向量元素时，会在下标中注明索引。例如，向量 $\mathbf{x}$ 的第 $0^{th}$ 个元素为 $x_0$。请注意，此处的 x 不是粗体。  

<a name="toc_40015_3.2"></a>
## 3.2 NumPy 数组

NumPy 的基本数据结构是可索引的 n 维*数组*，其中包含类型相同的元素（`dtype`）。你可能马上会注意到，我们赋予了“维度”一词另一层含义。上面，它表示向量中的元素数量；这里，维度表示数组的索引数量。一维或 1-D 数组有一个索引。在课程 1 中，我们会用 NumPy 一维数组表示向量。

 - 一维数组，形状为 (n,)：包含 n 个元素，索引从 [0] 到 [n-1]
 

<a name="toc_40015_3.3"></a>
## 3.3 创建向量

NumPy 中的数据创建例程通常将对象形状作为第一个参数。它既可以是表示一维结果的单个值，也可以是指定结果形状的元组 (n,m,...)。下面是使用这些例程创建向量的示例。

```python
# NumPy routines which allocate memory and fill arrays with value
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

一些数据创建例程不接收形状元组：

```python
# NumPy routines which allocate memory and fill arrays with value but do not accept shape as input argument
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

也可以手动指定值。

```python
# NumPy routines which allocate memory and fill with user specified values
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

这些方法都创建了一个包含四个元素的一维向量 `a`。`a.shape` 返回维度。这里可以看到 a.shape = `(4,)`，表示一个包含 4 个元素的一维数组。

## 3.4 向量运算
让我们探索一些向量运算。
<a name="toc_40015_3.4.1"></a>
### 3.4.1 索引
可以通过索引和切片访问向量中的元素。NumPy 提供了一套非常完整的索引和切片功能。这里仅探讨课程所需的基础知识。更多详细信息请参阅[切片和索引](https://NumPy.org/doc/stable/reference/arrays.indexing.html)。  
**索引**是指根据数组中元素的位置引用数组中的*一个元素*。  
**切片**是指根据索引从数组中获取元素的一个*子集*。  
NumPy 从零开始索引，因此向量 $\mathbf{a}$ 的第 3 个元素是 `a[2]`。

```python
#vector indexing operations on 1-D vectors
a = np.arange(10)
print(a)

#access an element
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# access the last element, negative indexes count from the end
print(f"a[-1] = {a[-1]}")

#indexs must be within the range of the vector or they will produce and error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

<a name="toc_40015_3.4.2"></a>
### 3.4.2 切片
切片使用一组三个值（`start:stop:step`）创建一个索引数组。只使用其中一部分值也有效。通过示例最容易说明其用法：

```python
#vector slicing operations
a = np.arange(10)
print(f"a         = {a}")

#access 5 consecutive elements (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# access 3 elements separated by two 
c = a[2:7:2];     print("a[2:7:2] = ", c)

# access all elements index 3 and above
c = a[3:];        print("a[3:]    = ", c)

# access all elements below index 3
c = a[:3];        print("a[:3]    = ", c)

# access all elements
c = a[:];         print("a[:]     = ", c)
```

<a name="toc_40015_3.4.3"></a>
### 3.4.3 单向量运算
有许多实用运算只涉及一个向量。

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# negate elements of a
b = -a 
print(f"b = -a        : {b}")

# sum all elements of a, returns a scalar
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
```

<a name="toc_40015_3.4.4"></a>
### 3.4.4 向量的逐元素运算
大多数 NumPy 算术、逻辑和比较运算也适用于向量。这些运算符逐元素执行运算。例如：
$$ \mathbf{a} + \mathbf{b} = \sum_{i=0}^{n-1} a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
```

当然，要使其正确工作，这些向量的大小必须相同：

```python
#try a mismatched vector operation
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

向量可以由标量值进行“缩放”。标量值就是一个数。标量会乘以向量中的所有元素。

<a name="toc_40015_3.4.6"></a>
### 3.4.6 向量-向量点积
点积是线性代数和 NumPy 中的核心运算。本课程会大量使用该运算，因此应充分理解。点积如下所示。

<img src="../work/images/C1_W2_Lab04_dot_notrans.gif" width=800>

点积将两个向量中的值逐元素相乘，然后对结果求和。
向量点积要求两个向量的维度相同。

下面，让我们实现自己的点积版本：

**使用 for 循环**实现一个返回两个向量点积的函数。给定输入 $a$ 和 $b$，该函数返回：
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
假设 `a` 和 `b` 的形状相同。

```python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

请注意，点积应返回一个标量值。

让我们使用 `np.dot` 尝试相同的运算。  

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

上面可以看到，一维情况下的结果与我们的实现相符。

<a name="toc_40015_3.4.7"></a>
### 3.4.7 速度至关重要：向量与 for 循环
我们使用 NumPy 库是因为它可以提高速度和内存效率。下面进行演示：

```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

因此，在这个示例中，向量化带来了显著的速度提升。这是因为 NumPy 能更好地利用底层硬件中可用的数据并行能力。GPU 和现代 CPU 实现了单指令多数据（SIMD）流水线，可以并行发出多个操作。这对机器学习至关重要，因为数据集通常非常庞大。

## 4.4 矩阵运算
让我们探索一些矩阵运算。

<a name="toc_40015_4.4.1"></a>
### 4.4.1 索引

矩阵包含第二个索引。两个索引分别表示 [行, 列]。访问操作可以返回一个元素，也可以返回一行/一列。请看下面：

```python
#vector indexing operations on matrices
a = np.arange(6).reshape(-1, 2)   #reshape is a convenient way to create matrices
print(f"a.shape: {a.shape}, \na= {a}")

#access an element
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

#access a row
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

最后一个示例值得特别注意。仅指定行来访问矩阵时，将返回一个*一维向量*。

**重塑**  
前面的示例使用了 [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) 来改变数组的形状。  
`a = np.arange(6).reshape(-1, 2) `   
这行代码首先创建了一个包含六个元素的*一维向量*。然后，它使用 reshape 命令将该向量重塑为一个*二维*数组。也可以写成：  
`a = np.arange(6).reshape(3, 2) `  
从而得到同一个 3 行 2 列的数组。
-1 参数指示该例程根据数组大小和列数来计算行数。

<a name="toc_40015_4.4.2"></a>
### 4.4.2 切片
切片使用一组三个值（`start:stop:step`）创建索引数组，也可以只使用其中一部分值。通过示例最容易理解其用法：

```python
#vector 2-D slicing operations
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#access 5 consecutive elements (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

#access 5 consecutive elements (start:stop:step) in two rows
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# access all elements
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# access all elements in one row (very common usage)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# same as
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
