## 正则化逻辑回归的代价函数

下面，你将实现正则化逻辑回归的代价函数。

回顾一下，对于正则化逻辑回归，代价函数的形式为
$$J(\mathbf{w},b) = \frac{1}{m}  \sum_{i=0}^{m-1} \left[ -y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \right] + \frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2$$

将其与不带正则化的代价函数（你已在之前的实验中实现）进行比较，后者的形式为

$$ J(\mathbf{w},b) = \frac{1}{m}\sum_{i=0}^{m-1} \left[ (-y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)\right]$$

区别在于正则化项，即 $$\frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2$$

请注意，我们不对参数 $b$ 进行正则化。

请完成 `compute_cost_reg` 函数，以执行以下操作：
- 与此前的 `compute_cost` 完全一样，计算训练集中所有样本的代价
    - 在循环外创建一个变量来存储代价
    - 遍历训练集中的所有样本。
        - 计算每个训练样本的代价
            - 计算 `z`
            - $ z =  w_0x_0^{(i)} + w_1x_1^{(i)} + b = \mathbf{x}^{(i)T}\mathbf{w} +b $
            - 预测 `f`，其中 `g` 是 sigmoid 函数
                - $ f =  g(z) $
            - 计算每个样本的代价
                $cost =  (-y^{(i)} * \log f) - ((1 - y^{(i)})\log( 1 - f))$
         - 将该代价添加到循环外创建的总代价变量中
    - 对所有迭代的代价求和，再将总和除以样本数。

        
- 计算正则化项：
- 在循环外创建一个正则化代价变量，用于累加正则化代价
- 遍历 `n`（特征数）
    - 计算 `w_j**2`
    - 将其添加到正则化代价变量中
    
- 将两个循环的输出相加，得到总和，即：
    - 所有训练样本的代价之和除以样本数
    - 遍历 $w$ 得到的正则化代价之和乘以 $\frac{\lambda}{2m}$

执行这些操作时，请记住变量 X 和 y 不是标量值，而是形状分别为（$m, n$）和（$𝑚$, 1）的矩阵，其中 $𝑛$ 是特征数，$𝑚$ 是训练样本数。

```python
def compute_cost_logistic_reg(X, y, w, b, lambda_ = 1):
    """
    Computes the cost over all examples
    Args:
      X : (array_like Shape (m,n)) data, m examples by n features
      y : (array_like Shape (m,1)) target value 
      w : (array_like Shape (n,1)) Values of parameters of the model      
      b : (array_like Shape (n,1)) Values of bias parameter of the model
      lambda_ : (scalar, float)    Controls amount of regularization
    Returns:
      total_cost: (scalar)         cost 
    """

    m, n = X.shape
    cost = 0.
    for i in range(m):
    ### START CODE HERE ### 
    ### BEGIN SOLUTION ###  
        z = X[i] @  w + b
        f = sigmoid(z)     
        cost +=  -y[i]*np.log(f) - (1-y[i])*np.log(1-f)
             
    total_cost = (1/m) * cost
    reg_cost = 0
    for j in range(n):
        reg_cost += (w[j]**2)
    
    total_cost = total_cost + (lambda_/(2*m))*reg_cost
    ### END SOLUTION ###  
    ### END CODE HERE ### 

    return total_cost[0]
```

运行下面的单元格，检查 `compute_cost_reg` 函数的实现。

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
initial_w = np.random.rand(X_tmp.shape[1]).reshape(-1,1)-0.5
initial_b = 0.5
lambda_ = 1
cost = compute_cost_logistic_reg(X_tmp, y_tmp, initial_w, initial_b, lambda_)

print("Regularized cost at initial w, b:", cost)
```
