### 非向量化版本

回顾一下，梯度下降的每次迭代都会执行以下更新：

$$w_j :=w_j - \alpha \frac{\partial J(w_j)}{\partial w}$$

对所有 $j$ 同时更新 $w_j$，其中：

$$ \frac{\partial J(w)}{\partial w_j}  = \frac{1}{m} \sum\limits_{i = 1}^{m} (f_{w}(x^{(i)}) - y^{(i)})x_j^{(i)}$$

- **注意**：虽然这个梯度看起来与线性回归梯度相同，但公式实际上不同，因为线性回归和逻辑回归对 $f_w(x)$ 的定义不同。

本实验将实现 $\frac{\partial J(w_j)}{\partial w}$。
* m 是数据集中的训练样本数量；

    
* $f_{w}(x^{(i)})$ 是模型的预测值，$y^{(i)}$ 是实际标签；

* 对于上面给定数据集的逻辑回归模型，可以将模型表示为：

    $f_{w}(x) = g(w_0 + w_1x_1 + w_2x_2)$

    其中 $g(z)$ 是 sigmoid 函数：

    $g(z) = \frac{1}{1+e^{-z}}$
    
    
* **预处理步骤**

   为了便于实现，我们将在 $X$ 中额外添加一列 1（作为 $x_0$），从而使：
    $f_{w}(x) = g(w_0x_0 + w_1x_1 + w_2x_2)$
    
    这样，在计算模型预测 $f_{w}(x)$ 时，就可以编写一个 for 循环，在每一步计算 $w_jx_j$。
