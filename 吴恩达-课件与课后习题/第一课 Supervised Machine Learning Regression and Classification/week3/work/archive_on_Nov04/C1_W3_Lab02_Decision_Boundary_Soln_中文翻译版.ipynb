## 逻辑回归模型


* 假设你想在这些数据上训练一个逻辑回归模型，其形式为：

  $f(x) = g(w_0x_0+w_1x_1 + b)$
  
  其中 $g(z) = \frac{1}{1+e^{-z}}$，即 sigmoid 函数


* 假设你训练了模型，并得到参数 $b = -3, w_0 = 1, w_1 = 1$。也就是说：

  $f(x) = g(x_0+x_1-3)$

  （你将在课程后面学习如何将这些参数拟合到数据）
  
  
让我们通过绘制决策边界，尝试理解这个训练好的模型在预测什么

### 逻辑回归与决策边界回顾

* 回想一下，对于逻辑回归，模型表示为

  $f_{\mathbf{w},b}(x) = g(x^Tw)$

  其中

  $g(z) = \frac{1}{1+e^{-z}}$
  
  $g(z)$ 称为 sigmoid 函数，它将所有输入值映射到 0 和 1 之间。
  

* 我们将模型输出（$f_{\mathbf{w},b}(x)$）解释为：在给定 $x$ 且由 $b$ 和 $w$ 参数化的条件下，$y=1$ 的概率。
* 因此，要从逻辑回归模型获得最终预测（$y=0$ 或 $y=1$），可以采用以下启发式规则：

  如果 $f_{\mathbf{w},b}(x) >= 0.5$，预测 $y=1$
  
  如果 $f_{\mathbf{w},b}(x) < 0.5$，预测 $y=0$
  
  
* 由于 $f_{\mathbf{w},b}(x) = g(x^Tw)$，让我们绘制 sigmoid 函数，看看 $g(z) >= 0.5$ 所在的位置

我们已经为你实现了 `sigmoid` 函数，只需按下面代码块所示导入并使用即可。

```python
import matplotlib.pyplot as plt
import numpy as np
from lab_utils import sigmoid


# Plot sigmoid(z) over a range of values from -20 to 20
z = np.arange(-20,20)
plt.plot(z, sigmoid(z), c="b")

# Plot and annotate the point where sigmoid(z) = 0.5
plt.plot(0, sigmoid(0), 'ro')
plt.text(0, sigmoid(0), '(x={}, y={})'.format(0, sigmoid(0)))
```

* 如你所见，当 $z >=0$ 时，$g(z) >= 0.5$

* 对于逻辑回归模型，$z = x^Tw$。因此，

  如果 $x^Tw >= 0$，模型预测为 $y=1$
  
  如果 $x^Tw < 0$，模型预测为 $y=0$
  
  
  
### 绘制决策边界

现在，让我们回到示例，理解逻辑回归模型如何进行预测。

* 逻辑回归模型的形式为

  $f(x) = g(-3 + x_0+x_1)$


* 根据上面学到的内容，可以看出，当 $-3 + x_0+x_1 >= 0$ 时，该模型预测 $y=1$

让我们通过图形来观察。首先绘制 $-3 + x_0+x_1 = 0$，它等价于 $x_1 = 3 - x_0$。

**练习**

请通过编写 `x1 = 3-x0` 来补全下面的代码块

```python
# Choose values between 0 and 6
x0 = np.arange(0,6)

### START CODE HERE ### 
### BEGIN SOLUTION ###  
x1 = 3 - x0
### END CODE HERE ###
### END SOLUTION ###  
# Plot the decision boundary
plt.plot(x0,x1, c="b")
plt.axis([0, 6, 0, 6])

# Fill the region below the line
plt.fill_between(x0,x1, alpha=0.2)

# Plot the original data
plot_data(X,y)
# Set the y-axis label
plt.ylabel('x1')
# Set the x-axis label
plt.xlabel('x0')
```

* 在上图中，蓝线应表示直线 $x_0 + x_1 - 3 = 0$，它应在 3 处与 x1 轴相交（如果设置 $x_1$ = 3，则 $x_0$ = 0），并在 3 处与 x0 轴相交（如果设置 $x_1$ = 0，则 $x_0$ = 3）。


* 阴影区域表示 $-3 + x_0+x_1 < 0$。直线上方的区域为 $-3 + x_0+x_1 > 0$。


* 因此，对于该模型，阴影区域（直线下方）中的任何点都被分类为 $y=0$，而直线上或直线上方的任何点都被分类为 $y=1$。这条线称为“决策边界”。
