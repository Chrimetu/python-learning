## 逻辑回归模型


* 假设你想在这些数据上训练一个逻辑回归模型，其形式为

  $f(x) = g(w_0 + w_1x_1+w_2x_2)$

* 假设你训练了模型，并得到参数 $w_0 = -3, w_1 = 1, w_2 = 1$。也就是说，

  $f(x) = g(-3 + x_1+x_2)$

让我们绘制模型的决策边界，尝试理解这个训练后的模型会预测什么。

### 逻辑回归与决策边界复习

* 回顾一下，对于逻辑回归，模型表示为

  $f(x) = g(w^Tx)$

  其中

  $g(z) = \frac{1}{1+e^{-z}}$
  
  $g(z)$ 称为 sigmoid 函数，它将所有输入值映射到 0 和 1 之间。
  

* 我们将模型的输出（$f(x)$）解释为：在给定 $x$ 且由 $w$ 参数化时，$y=1$ 的概率。
* 因此，要从逻辑回归模型得到最终预测（$y=0$ 或 $y=1$），可以使用以下启发式规则：

  如果 $f(x) >= 0.5$，预测 $y=1$
  
  如果 $f(x) < 0.5$，预测 $y=0$
  
  
* 由于 $f(x) = g(w^Tx)$，让我们绘制 sigmoid 函数，看看 $g(z) >= 0.5$ 位于何处。

我们已经为您实现了 `sigmoid` 函数，您只需像下面代码块所示那样导入并使用它即可。

```python
import matplotlib.pyplot as plt
import numpy as np
from lab_utils import sigmoid


# Plot sigmoid(z) over a range of values from -20 to 20
z = np.arange(-20,20)
plt.plot(z, sigmoid(z), c="b")

# Plot and annotate the point where sigmoid(z) = 0.5
plt.plot(0, sigmoid(0), 'ro')
plt.text(0, sigmoid(0), '(x={}, y={})'.format(0, sigmoid(0)))
```

* 可以看到，$g(z) >= 0.5$ 对于 $z >=0$

* 对于逻辑回归模型，$z = w^Tx$。因此：

  如果 $w^Tx >= 0$，模型预测 $y=1$
  
  如果 $w^Tx < 0$，模型预测 $y=0$
