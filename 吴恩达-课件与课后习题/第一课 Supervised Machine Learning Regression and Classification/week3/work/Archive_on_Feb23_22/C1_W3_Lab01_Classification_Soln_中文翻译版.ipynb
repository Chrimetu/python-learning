## 分类问题
<img align="left" src="./images/C1_W3_Classification.png"     style=" width:380px; padding: 10px; " > 分类问题的例子包括：将电子邮件识别为垃圾邮件或非垃圾邮件，或者判断肿瘤是恶性还是良性。具体来说，这些都是具有两种可能结果的*二分类*示例。结果可以描述为成对的“正”/“负”，例如“是”/“否”、“真”/“假”或“1”/“0”。

分类数据集的图通常使用符号表示样本的结果。在下图中，“X”表示正值，“O”表示负结果。

```python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])
X_train2 = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train2 = np.array([0, 0, 0, 1, 1, 1])
```

```python
pos = y_train == 1
neg = y_train == 0

fig,ax = plt.subplots(1,2,figsize=(8,3))
#plot 1, single variable
ax[0].scatter(x_train[pos], y_train[pos], marker='x', s=80, c = 'red', label="y=1")
ax[0].scatter(x_train[neg], y_train[neg], marker='o', s=100, label="y=0", facecolors='none', edgecolors=dlblue,lw=3)

ax[0].set_ylim(-0.08,1.1)
ax[0].set_ylabel('y', fontsize=12)
ax[0].set_xlabel('x', fontsize=12)
ax[0].set_title('one variable plot')
ax[0].legend()

#plot 2, two variables
plot_data(X_train2, y_train2, ax[1])
ax[1].axis([0, 4, 0, 4])
ax[1].set_ylabel('$x_1$', fontsize=12)
ax[1].set_xlabel('$x_0$', fontsize=12)
ax[1].set_title('two variable plot')
ax[1].legend()
plt.tight_layout()
plt.show()
```

请注意上图中的以下几点：
- 在单变量图中，正类结果同时以红色“X”和 y=1 表示。负类结果以蓝色“O”表示，并位于 y=0。
   - 回顾一下，在线性回归中，y 不会局限于两个值，而可以取任意值。
- 在双变量图中，无法显示 y 轴。正类结果也用红色“X”表示，负类结果用蓝色“O”表示。
    - 回顾一下，在多变量线性回归中，y 不会局限于两个值，类似的图将是三维图。
