## Sigmoid 或逻辑函数

- 但是，由于输出变量 $y$ 只能是 0 或 1，我们希望分类模型的预测值位于 0 和 1 之间。

## Sigmoid 函数的公式

sigmoid 函数的公式如下：  

$g(z) = \frac{1}{1+e^{-z}}\tag{1}$

在逻辑回归中，z（sigmoid 函数的输入）是线性回归模型的输出。
- 对于单个样本，$z$ 是标量。
- 对于多个样本，$z$ 可以是由 $m$ 个值组成的向量，每个样本对应一个值。
- sigmoid 函数的实现应兼容这两种潜在的输入格式。
让我们用 Python 来实现。

NumPy 提供了一个名为 [`exp()`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html) 的函数，可以方便地计算输入数组（`z`）中所有元素的指数（$e^{z}$）。
 
如下所示，它也可以接收单个数字作为输入。

```python
# Input is an array. 
input_array = np.array([1,2,3])
exp_array = np.exp(input_array)

print("Input to exp:", input_array)
print("Output of exp:", exp_array)

# Input is a single number
input_val = 1  
exp_val = np.exp(input_val)

print("Input to exp:", input_val)
print("Output of exp:", exp_val)
```

`sigmoid` 函数的 Python 实现如下方单元格所示。

```python
def sigmoid(z):
    """
    Compute the sigmoid of z

    Args:
        z (ndarray): A scalar, numpy array of any size.

    Returns:
        g (ndarray): sigmoid(z), with the same shape as z
         
    """

    g = 1/(1+np.exp(-z))
   
    return g
```

让我们看看该函数在 `z` 取不同值时的输出。

```python
# Generate an array of evenly spaced values between -10 and 10
z_tmp = np.arange(-10,11)

# Use the function implemented above to get the sigmoid values
y = sigmoid(z_tmp)

# Code for pretty printing the two arrays next to each other
np.set_printoptions(precision=3) 
print("Input (z), Output (sigmoid(z))")
print(np.c_[z_tmp, y])
```

左列中的值是 `z`，右列中的值是 `sigmoid(z)`。可以看到，sigmoid 的输入值范围为 -10 到 10，输出值范围为 0 到 1。

现在，尝试使用 `matplotlib` 库绘制此函数。

```python
# Plot z vs sigmoid(z)
fig,ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(z_tmp, y, c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```

可以看到，当 `z` 趋向较大的负值时，sigmoid 函数趋近于 `0`；当 `z` 趋向较大的正值时，sigmoid 函数趋近于 `1`。

## 逻辑回归
<img align="left" src="./images/C1_W3_LogisticRegression_right.png"     style=" width:300px; padding: 10px; " > 逻辑回归模型将 sigmoid 应用于熟悉的线性回归模型，如下所示：

$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot \mathbf{x}^{(i)} + b ) \tag{2} $$ 

  其中：

  $g(z) = \frac{1}{1+e^{-z}}\tag{3}$
